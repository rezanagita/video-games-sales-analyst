#Business Understanding

Data tersebut menunjukkan data penjualan video  game berdasarkan rank (menunjukkan peminat terbanyak), nama game, platform, genre, publisher, tahun rilis, dan penjualan di berbagai wilayah di dunia (NA_Sales menunjukkan penjualan di wilayah North America atau Amerika Utara, EU_Sales menunjukkan penjualan di wilayah Europe atau Eropa, JP_Sales menunjukkan penjualan di wilayah Japan atau Jepang, Other_Sales menunjukkan penjualan di wilayah lain, dan Global_Sales menunjukkan penjualan secara global). Satuan total penjualan video game dalam satuan juta unit sehingga jumlah penjualan tersebut ditampilkan dalam berapa juta copy yang sudah terjual. 0,01 menunjukkan 10 ribu juta copy yang sudah terjual.

#Python Analysis
1. Tampilkan 10 Global_Sales teratas berdasarkan name, genre, platform
2. Tampilkan 10 platfrom yang paling banyak digunakan
3. Tampilkan berapa banyak video game di setiap genre
4. Tampilkan 10 video game yang terbanyak terjual di wilayah Amerika Utara berdasarkan nama dan genre
5. Tampilkan 10 video game yang terbanyak terjual di wilayah Eropa berdasarkan nama dan genre
6. Tampilkan 10 video game yang terbanyak terjual di wilayah Jepang berdasarkan nama dan genre
7. Tampilkan 10 teratas dari total video game yang dikeluarkan publisher
8. Tampilkan 10 teratas dari penjualan video game di Amerika Utara, Eropa, dan Jepang
9. Tampilkan Global_Sales setiap tahunnya
10. Tampilkan jumlah video game di setiap genre pada tahun 2008

```python
#1. Tampilkan 10 Global_Sales teratas berdasarkan name, genre, platform
game_sales = (
    df.groupby(["Name", "Genre", "Publisher"])["Global_Sales (millions_copy)"]
    .sum()
    .reset_index()
    .sort_values(by="Global_Sales (millions_copy)", ascending=False)
    .reset_index(drop=True) #Merapikan nomor indeks tabel
)

game_sales.head(10)
```

```python
#2. Tampilkan 10 platfrom yang paling banyak digunakan
platform_game_count = (
    df.groupby("Platform")["Name"]
    .nunique()
    .reset_index(name="Jumlah_Video_Game")
    .sort_values(by="Jumlah_Video_Game", ascending=False)
    .reset_index(drop=True) #Merapikan nomor indeks tabel
)

platform_game_count.head(10)
```

```python
#3. Tampilkan berapa banyak video game di setiap genre
genre_game_count = (
    df.groupby("Genre")["Name"]
    .nunique()
    .reset_index(name="Jumlah_Video_Game")
    .sort_values(by="Jumlah_Video_Game", ascending=False)
    .reset_index(drop=True) #Merapikan nomor indeks tabel
)

genre_game_count
```

```python
#4. Tampilkan 10 video game yang terbanyak terjual di wilayah Amerika Utara berdasarkan nama dan genre
top_10_na = (
    df.groupby(["Name", "Genre"])["NA_Sales (millions_copy)"]
    .sum()
    .reset_index(name="Total_NA_Sales")
    .sort_values(by="Total_NA_Sales", ascending=False)
    .head(10)
    .reset_index(drop=True) #Merapikan nomor indeks tabel
)

top_10_na
```

```python
#5. Tampilkan 10 video game yang terbanyak terjual di wilayah Eropa berdasarkan nama dan genre
top_10_eu = (
    df.groupby(["Name", "Genre"])["EU_Sales (millions_copy)"]
    .sum()
    .reset_index(name="Total_EU_Sales")
    .sort_values(by="Total_EU_Sales", ascending=False)
    .head(10)
    .reset_index(drop=True) #Merapikan nomor indeks tabel
)

top_10_eu
```

```python
#6. Tampilkan 10 video game yang terbanyak terjual di wilayah Amerika Utara berdasarkan nama dan genre
top_10_jp = (
    df.groupby(["Name", "Genre"])["JP_Sales (millions_copy)"]
    .sum()
    .reset_index(name="Total_JP_Sales")
    .sort_values(by="Total_JP_Sales", ascending=False)
    .head(10)
    .reset_index(drop=True) #Merapikan nomor indeks tabel
)

top_10_jp
```

```python
#7.Tampilkan 10 teratas dari total video game yang dikeluarkan publisher
publisher_game_count = (
    df.groupby("Publisher")["Name"]
    .nunique()
    .reset_index(name="Total_Video_Game")
    .sort_values(by="Total_Video_Game", ascending=False)
    .head(10)
    .reset_index(drop=True) #Merapikan nomor indeks tabel
)

publisher_game_count
```

```python
#8.Tampilkan 10 teratas dari penjualan video game di Amerika Utara, Eropa, dan Jepang
df["Penjualan di Amerika Utara, Eropa, dan Jepang (millions_copy)"] = df["Global_Sales (millions_copy)"] - df["Other_Sales (millions_copy)"]

top_10_combined_sales = (
    df.groupby("Name")["Penjualan di Amerika Utara, Eropa, dan Jepang (millions_copy)"]
    .sum()
    .reset_index(name="Total_Sales_NA_EU_JP")
    .sort_values(by="Total_Sales_NA_EU_JP", ascending=False)
    .head(10)
    .reset_index(drop=True) #Merapikan nomor indeks tabel
)

top_10_combined_sales
```

```python
#9. Tampilkan Global_Sales setiap tahunnya
yearly_sales = (
    df.groupby('Year')['Global_Sales (millions_copy)']
    .sum()
    .reset_index()
    .sort_values(by='Global_Sales (millions_copy)',ascending=False)
    .reset_index(drop=True) #Merapikan nomor indeks tabel
)

yearly_sales
```

```python
#10. Tampilkan jumlah video game di setiap genre pada tahun 2008
Video_Game_2008 = (
    df[df['Year'] == 2008]
    .groupby('Genre')['Name']
    .nunique()
    .reset_index(name="Jumlah Video Game")
    .sort_values(by="Jumlah Video Game", ascending=False)
    .reset_index(drop=True)
)

print(Video_Game_2008)
```
