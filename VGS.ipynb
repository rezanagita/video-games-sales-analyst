#Business Understanding

Data tersebut menunjukkan data penjualan video  game berdasarkan rank (menunjukkan peminat terbanyak), nama game, platform, genre, publisher, tahun rilis, dan penjualan di berbagai wilayah di dunia (NA_Sales menunjukkan penjualan di wilayah North America atau Amerika Utara, EU_Sales menunjukkan penjualan di wilayah Europe atau Eropa, JP_Sales menunjukkan penjualan di wilayah Japan atau Jepang, Other_Sales menunjukkan penjualan di wilayah lain, dan Global_Sales menunjukkan penjualan secara global). Satuan total penjualan video game dalam satuan juta unit sehingga jumlah penjualan tersebut ditampilkan dalam berapa juta copy yang sudah terjual. 0,01 menunjukkan 10 ribu juta copy yang sudah terjual.

# Business Matrix and Questions
1. Berapa Global sales yang dihasilkan berdasarkan name, genre, Publisher
2. platfrom yang paling banyk digunakan
3. tren genre yang paling populer/laku
4. video game nama, genre yang banyak terjual di wilayah amerika utara,
5. video game nama, genre yang banyak terjual di wilayah eropa
6. video game nama, genre yang banyak terjual di wilayah jepang
7. jumlah total penjualan berdasarkan publiser mengeluarkan game
8. Berapa total penjualan di 3 negara besar (amerika, eropa, jepang)

```python
# Mengubah nama kolom sales (coppy juta)
data.rename(columns={'NA_Sales': 'NA_Sales (millions_copy)'}, inplace=True)
data.rename(columns={'EU_Sales': 'EU_Sales (millions_copy)'}, inplace=True)
data.rename(columns={'JP_Sales': 'JP_Sales (millions_copy)'}, inplace=True)
data.rename(columns={'Other_Sales': 'Other_Sales (millions_copy)'}, inplace=True)
data.rename(columns={'Global_Sales': 'Global_Sales (millions_copy)'}, inplace=True)
```

```python
# 1 10 data global sales paling banyak
data[["Global_Sales (millions_copy)", "Name","Genre" ,"Publisher"]].head(5)
```

```python
# 2. platfrom
data[["Platform" ]].value_counts().head(10)
```

```python
# 3. tren genre yang paling populer/laku terjual semua tahun
data['Genre'].value_counts()
```

```python
# Menampilkan semua data
pd.set_option('display.max_rows', None)
pd.set_option('display.max_columns', None)
```

```python
# 4. video game nama, genre yang banyak terjual di wilayah amerika utara
data[["Name", "Genre", "NA_Sales (millions_copy)"]].sort_values(by="NA_Sales (millions_copy)", ascending=False)
```

```python
# 5. video game nama, genre yang banyak terjual di wilayah eropa
data[["Name", "Genre", "EU_Sales (millions_copy)"]].sort_values(by="EU_Sales (millions_copy)", ascending=False)
```

```python
# 6. video game nama, genre yang banyak terjual di wilayah jepang
data[["Name", "Genre", "JP_Sales (millions_copy)"]].sort_values(by="JP_Sales (millions_copy)", ascending=False)
```

```python
# 7. Berapa total video game  yang dikeluarkan oleh Publisher?
data.groupby(["Publisher"])["Name"].count().sort_values(ascending=False).head(10)
```

```python
# 8. Jumlah genre berdasarkan global sales
data.groupby('Genre')['Global_Sales (millions_copy)'].sum().sort_values(ascending=False)
```

```python
#8.Jumlah penjualan video game di Amerika Utara, Eropa, dan Jepang
data["Penjualan di Amerika Utara, Eropa, dan Jepang (millions_copy)"] = data["Global_Sales (millions_copy)"] - data["Other_Sales (millions_copy)"]

data[["Name", "Penjualan di Amerika Utara, Eropa, dan Jepang (millions_copy)"]].head(5)
```

```python
# (Tambahan) Groupby tahun, sum penjualan global
yearly_sales = data.groupby('Year')['Global_Sales (millions_copy)'].sum().reset_index().sort_values(by='Global_Sales (millions_copy)',ascending=False)

yearly_sales
```

```python
# Filter data tahun 2008, lalu ambil kolom 'Genre'
genre_2008 = data[data['Year'] == 2008]['Genre'].count('Global_Sales (millions_copy)')
genre_2008.head()
```
